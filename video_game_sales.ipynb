```python
missing_names = df.loc[df["Year"].isna(), "Name"].unique()

year_info = (
    df[df["Name"].isin(missing_names)]
    .groupby("Name")["Year"]
    .agg(["count", "nunique", "unique"])
)

candidates = year_info[
    (year_info["count"] > 0) &
    (year_info["nunique"] == 1)
]

print("Araştırmaya değer oyun sayısı:", len(candidates))
print(candidates)
```

```python
candidate_names = candidates.index

candidate_rows = df.loc[
    df["Name"].isin(candidate_names),
    ["Name", "Platform", "Year", "Genre", "Publisher"]
].sort_values(["Name", "Platform"])

candidate_rows
```

```python
platform_year_counts = (
    df.groupby(["Name", "Platform"])["Year"]
    .count()
)

print("Yılı bilinen Name + Platform grubu sayısı:",
      (platform_year_counts > 0).sum())
```

```python
group = df.groupby(["Name", "Platform"])["Year"]

year_count = group.transform("count")
year_nunique = group.transform("nunique")
year_first = group.transform("first")

mask = (
    df["Year"].isna()
    & (year_count > 0)
    & (year_nunique == 1)
)

print("Güvenilir eşleşme adayı:", mask.sum())

df.loc[mask, ["Name", "Platform", "Year"]].head(20)
df.loc[mask, "Year"] = year_first[mask] 
df.loc[
    df["Name"].eq("Sonic the Hedgehog"),
    ["Name", "Platform", "Year"]
]
```

```python
missing_names = df.loc[df["Year"].isna(), "Name"].unique()

year_info = (
    df[df["Name"].isin(missing_names)]
    .groupby("Name")["Year"]
    .agg(["count", "nunique", "unique"])
)

cross_platform_candidates = year_info[
    (year_info["count"] > 0) &
    (year_info["nunique"] == 1)
]

print("Farklı platformlardan araştırılacak oyun sayısı:",
      len(cross_platform_candidates))
```

```python
missing_names = df.loc[df["Year"].isna(), "Name"].unique()

cross_platform = df[df["Name"].isin(missing_names)]

cross_platform_summary = (
    cross_platform.groupby("Name")
    .agg(
        Platforms=("Platform", lambda x: list(x)),
        Years=("Year", lambda x: list(x)),
        Publishers=("Publisher", lambda x: list(x))
    )
)

cross_platform_summary.loc[
    cross_platform_summary.index.isin(cross_platform_candidates.index)
].head(20)
```
